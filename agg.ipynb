```python
# Группировка 
df.groupby(['country']).sum() # Получили дф с индексом 'Страна' и значние - это сумма таких то значений по стране
# Сюда можно применять все те же функции агрегации
# df.groupby(['ad_channel'])[['impressions']].sum()
# df.groupby(['ad_channel']).agg( # В функцию agg запихиваем словарь, где ключ это название колонки, а значение это список из функций, которых мы хотим применить к колонке
#     {'impressions': ['sum', 'count'],
#      'ad_spend': ['sum', 'mean']}
# )
```

```python
df.groupby(['ad_channel']).agg(
    total_impressions = ('impressions', 'sum'), # Здесь называем колонки как мы хотим, и туда передаем кортеж из 1 - колонка, 2 - агрегирующая функция
    avg_daily_spend = ('impressions', 'mean')
)
```

```python
rev_per_click ={
    'RU': 0.04,
    'UA': 0.35,
    'KZ': 0.4,
    'US': 0.07,
    'CN': 0.5
}

def calculate_roi(group): # Передаем в функцию наш маленький дф(группу)
    country = group.reset_index()['country'].iloc[0] # Взяли 1-ое значение какой-то колонки и присвоили переменной 
    clicks = group['clicks'].sum() # Далее агрегировали колонку нашего маленького дф - просуммировали и записали в переменную
    ad_spend = group['ad_spend'].sum() # То же самое
    revenue = clicks * rev_per_click[country] # Умножили клики на rpc каждой страны в нашем словаре 
    roi = revenue / ad_spend # Результат
    return roi # Вернули результат


df.groupby('country').apply(calculate_roi) # Здесь берем наш дф, группируем его по определенному значению(в нашем случаем - это страна(бывшая колонка))
# Далее применяем нашу кастомную функцию к нашему сгруппированому дф
# Условно, group - это маленький датафрейм 
```

```python
# Сортировка 
# .sort_values() # Сорировка по значениям 
# .sort_index() # Сортировка по индексам
df.groupby('country').apply(calculate_roi).sort_values(ascending=False)
```
