```python
import pandas as pd
```

```python
orders = pd.read_csv('practice/orders.csv')
orders.head()
```

```python
costs = pd.read_csv('practice/costs.csv')
costs.head()
```

```python
daily_revenue = orders.groupby('order_created_dt').agg(revenue = ('order_amount', 'sum'))
daily_revenue.head()
```

```python
daily_spend = costs.groupby('date').agg(spend = ('cost', 'sum'))
daily_spend.head()
```

```python
(
    daily_revenue
    .merge(daily_spend, how='outer', left_index=True, right_index=True)
    .fillna(0)
).head(50)
# У нас получился дф, где иногда дневная выручка = 0, а иногда дневные траты = 0. Именно поэтому здесь подходит полный джоин, остальные джоины не включили бы эти дни в свои датафреймы 
```
