```python
import numpy as np
import pandas as pd
```

```python
# Массив данных
x = [1, 2, 3]
t = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

items = [2, 5, 15]
prices = [45, 22, 90]
```

```python
# Создание массива(одномерного), у np массива у каждого элемента есть свой порядковый номер(он не меняется, это обязательно должны быть числа и в порядке)
a_items = np.array(items)
a_prices = np.array(prices)
print(a_items * a_prices)
```

```python
# Series - одномерный массив данных, в отличие от массива Numpy, в pd Series индексы могут меняться и быть строками или любыми другими значениями 
pd_items = pd.Series(items)
print(pd_items)


```

```python
# DataFrame - двумерный массив данных(или Таблица), По сути DataFrame состоит из Серий
df_t = pd.DataFrame(t)
print(df_t)
```

```python
books = [
    {'title': 'foobar',
     'pub_date': '2023-01-01',
     'author': 'John Smith'},
     {'title': 'foo',
     'pub_date': '2023-01-02',
     'author': 'John Smith'},
     {'title': 'bar',
     'pub_date': '2023-01-03',
     'author': 'Jane Smith'}
]
df = pd.DataFrame(books)
print(df)
```
