```python
# Интерфейс - набор методов, с помощью которых пользователь может решить свою задачу.
# Интрефейс прячет от пользователя ненужные ему элементы системы.

# API (application programming interface) - интерфейс, с помощью которого приложением может пользоваться другое приложение.
# Примеры API:
# Онлайн-брокер продает и покупает акции на бирже
# Яндекс.Такси узнает у Навигатора о пробках, чтобы назначить стоиомость поездки

# Как работают API. Протокол HTTP
# Протокол - набор правил, по которым устройства общаются между собой.
# HTTP(hypertext transfer protocol) - протокол обмена гипертекстом
# Гипертекст - текст со <a href="google.com">ссылками</a>

# Клиент-серверная аритектура 
# Клиент кликает на кнопку > тем самым делает запрос на сервер > сервер отдает ответ
# Слово сервер от слова serve(служить, отдавать)

# Структура HTTP-сообщения # Когда мы делаем запрос - это HTTP сообщение, когда сервер отправляет ответ - это тоже HTTP сообщение.
# Запрос: 
# Start Line:
            # Method: Нужное действие, # Чаще всего - либо хотим получить какую-либо информацию, либо отправить
            # URI(Uniform Resource Identifier): адрес обращения 
# Headers:
         # Метаданные запроса: формат, язык и т.п. # Например csv или json, на каком языке они или сколько строк всего
         # Также там могут быть ключи авторизации
# Body:
      # Данные, которые мы хотим передать # Например список пользователей, которые когда-то что-то купили у нас

# Ответ:
# Start Line:
            # Status Code: код ответа (200 OK) 
# Headers:
         # Метаданные запроса: формат, язык и т.п. # Например csv или json, на каком языке они или сколько строк всего
         # Также там могут быть ключи авторизации
# Body:
      # Данные, которые мы запрашивали # Например список пользователей, которые когда-то что-то купили у нас

# HTTP-методы:
# 1) GET - "Дайте данных, пожалуйста"
# 2) POST - "Вот вам данные, положите куда-нибудь, чтобы не потерялось"
# Еще есть DELETE, PUT, PATCH.

# Status Codes
# 1xx - процесс идет
# 2xx - все ок
# 3xx - редиректы(Перенаправили ваш запрос к другому ресурсу)
# 4xx - ошибка на стороне клиента
# 5xx - ошибка на стороне сервера

# Любая работа с API начинается с документации.
```
