```python
df = pd.read_csv('data_orders.csv', index_col='order_id')
df.head()
```

```python
# Редактирование колонок 
# df['price'] = df['price'] * 100 # Указыаем колонку и присваеваем новые значения путем какой-нибудь операции
# df['price'] = 0 # То же самое, распространится на все ячейки 
```

```python
# Редактивроание строк
df.loc[1001] = {'date': '2025-12-31', 'quantity': 3, 'region': 'East', 'product': 'fff', 'price': 1, 'category': 'vvv'} # Указали индекс и передали в него словарь с отредактированными данными, где указывали название колонки и новое значение 
df.loc[1001]
```

```python
# Редактирование ячеек
df.at[1002, 'date'] = None # Указали ячейку и записали туда новое значение 
```

```python
# Редактирование по фильтру 
df.loc[df['quantity'] == 1, 'price'] = df.loc[df['quantity'] == 1, 'price'] * .7 # К отфильтрованной колонки присвоили новые значения, которые получили с помощью операции
```

```python
# Добавление колонок 
df['foo'] = 1 # Добовляем новую колонку просто указывая новое имя колонки в дф и присваиваем значения 
df['price_discount'] = df['price'] * .7 # Или делаем новую колонку путем выполнения какой-то операции от другой колонки
df
```

```python
# Добавление строк
df.loc[1008] = {} # Указываем новый индекс и присваем либо пустоту, либо указыаем названия колонок и значения 
df
```

```python
# Удаление колонок и строк # По сути название и любой строки и любой колонки это - индекс(какой-то)
# df = df.drop('foo', axis='columns') # Указыаем индекс и по какой оси мы хотим сделать удаление
df = df.drop(1008, axis='rows')
# Если нужно удалить несколько строк или колонок, то передаем индексы списком 
```

```python
df = df.drop_duplicates('date', keep='first') # Так удаляются дубликаты в определенной колонки
```

```python
# Обработка пустых записей
df.dropna() # Удаляем строки с пустыми записями 
df.fillna('foo') # Либо передаем этим строкам какое-то значение 
```

```python
# Обработка датафреймов с помощью функции 'apply()'
# > 1 -> .5
# > 4 -> .3
# > 5 -> .1

def apply_discount(row):
    if row['quantity'] < 1:
        row['quantity'] *= .5
    elif row['quantity'] < 4:
        row['quantity'] *= .3
    else:
        row['quantity'] *= .1
    return row

# test = pd.DataFrame([{'quantity': 7}, {'quantity': 3}, {'quantity': 0.5}])
# # test
# test.apply(apply_discount, axis='columns')
df.head()
# df.apply(apply_discount, axis='columns')
```

```python
df.apply(apply_discount, axis='columns')
```
