```python
import pandas as pd
```

```python
df = pd.read_csv('daily_visitors_dynamic.csv')
df.head(10)
```

```python
# rolling()
df.groupby('date')['daily_visitors'].sum().head(10).rolling(7).mean() # Функция rolling() - Скользящая средная (Rolling Mean / Moving Average) вычисляет среднее значение временного ряда в пределах скользящего окна заданного размера. Это скользящее окно перемещается вдоль временного ряда и в каждой точке времени вычисляет среднее значение точек данных внутри этого окна.
# Прикол роллинга именно в окне, он может выбрать временной окно и посчиать значение за это окно
```

```python
# transform() # В то время как агрегирующая функция должна возвращать сокращенную версию данных, преобразование может вернуть версию полного набора данных, преобразованную ради дальнейшей их перекомпоновки. При подобном преобразовании форма выходных данных совпадает с формой входных.
# transform сохраняет формат исходного набора данных. Это уникальная особенность transform!

df['total_visitors'] = df.groupby('ad_channel')['daily_visitors'].transform(lambda x: sum(x))
df
```

```python
df['rolling_average'] = df.groupby('ad_channel')['daily_visitors'].transform(lambda x: x.rolling(7).mean())
df.head(10)
```

```python
df.sort_values(['ad_channel', 'date']).head(20)
```

```python
df['cumulative_sum'] = df.groupby('ad_channel')['daily_visitors'].transform(lambda x: x.expanding().sum())
df.sort_values(['ad_channel', 'date']).head(20)
# А прикол expandinga в том, что он берет и с самого начала накапливает значения в себе - Собирательный эффект
```

```python
# shift - По сути помогает нам получить данные в новых строках, взятые из предыдущих или следующих строк 
# foo = df.head()[['daily_visitors']]
# foo['yesterday_visitors'] = foo['daily_visitors'].shift(1)
# foo['result'] = foo['daily_visitors'] - foo['yesterday_visitors']
# foo

df['yesterday_visitors'] = df.groupby('ad_channel')['daily_visitors'].shift(1)
df['result'] = df['daily_visitors'] - df['yesterday_visitors']
df.sort_values(['ad_channel', 'date']).head(20)
df
```

```python
# DZ 1
df['Rolling_Avg'] = df['Stock_Price'].rolling(5).mean().round(2)
print(df.to_string(index=None))
```

```python
# DZ 2
df['Average_Sales'] = df.groupby('Region')['Sales'].transform(lambda x: x.mean())
df['Normalized_Sales'] = df['Sales'] / df['Average_Sales']
print(df.round(2).to_string(index=None))
```

```python
# DZ 3
df['Cumulative_Avg_Sales'] = df['Sales'].transform(lambda x: x.expanding().mean()).round(2)
print(df.to_string(index=None))
```

```python
# DZ 4
df['Daily_Change'] = df['Sales'] - df.groupby('Product_Name')['Sales'].shift(1) 
print(df.to_string(index=None))
```
