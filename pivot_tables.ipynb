```python
import pandas as pd
```

```python
df = pd.read_csv('ads.csv')
df.head(30)
```

```python
# Расходы на каждый рекламный канал в каждой стране 
# Стоимость клика в каждой стране и каждом рекламном канале 
# Данные были читаемыми 
```

```python
df.pivot_table( # Функция создает свойдну таблицу, указываем ей как хотим, чтобы выглядели строки, колонки, значения и какие функции к ним применить
    index='country',
    columns='ad_channel',
    values=['ad_spend', 'clicks'],
    aggfunc='sum'
).fillna(0) # Функция вместо NaN подставляет 0 в данном случае или можем удалить .dropna()
```

```python
df.set_index('country').loc['US'][['date', 'ad_channel', 'impressions']]
```

```python
# Функция .pivot(index=, columns=, values=) # По сути нужна чтобы переоформить наш дф в другой вид(Больше строк, меньше колонок или наоборот) 
# От пивот тейбл отличается отсутствием параметра агрегации
(
    df
    .set_index('country')
    .loc['US']
    .pivot(
                        index='date',
                        columns='ad_channel',
                        values='impressions'
)
    .fillna(0)
    .astype('int') # Меняет тип даных в значениях
)
```

```python
avg_spend = df.groupby(['country', 'ad_channel'])['ad_spend'].sum().mean()

def get_av(values):
    x = sum(values)
    if x < avg_spend:
        return 'below av'
    elif x == avg_spend:
        return 'av'
    else:
        return 'above_av'
    

    



# df.pivot_table( 
#     index='country',
#     columns='ad_channel',
#     values='ad_spend',
#     aggfunc=get_av
# )

# below av
# av
# above av
```

```python
# DZ 1
print(df.pivot(index='Product', columns='Month', values='Sales').fillna(0))
```

```python
# DZ 2
print(round(df.pivot_table(index='Product', columns='Month', values='Sales', aggfunc='mean'), 2))
```

```python
# DZ 3
print(df.pivot_table(index=['Rep', 'Product'], columns=None, values='Sales', aggfunc='sum'))
```

```python
# DZ 4
def calcul(group):
    s_sales = group['Sales'].sum()
    c_sales = group['Sales'].count()
    result = s_sales / c_sales
    return result
    
print(round(df.groupby('Product').apply(calcul), 2))
```
